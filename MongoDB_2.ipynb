##  Part 1: Theoretical Questions

### 1. What are the key differences between SQL and NoSQL databases?
* **Data Model:** SQL databases are relational and use structured tables with rows and columns (schema-based). NoSQL databases are non-relational and use flexible models like documents (JSON/BSON), key-value pairs, graphs, or wide-column stores.
* **Schema:** SQL requires a predefined schema where all rows in a table must follow the same structure. NoSQL is dynamic and schema-less, allowing different documents in the same collection to have different fields.
* **Scalability:** SQL databases are primarily vertically scalable (scaling up by upgrading hardware). NoSQL databases are designed for horizontal scalability (scaling out by adding more servers/nodes).
* **Transactions:** SQL databases generally follow strict ACID properties natively across complex multi-row operations. NoSQL traditionally focused on high throughput and eventual consistency (though modern MongoDB supports multi-document ACID transactions).

### 2. What makes MongoDB a good choice for modern applications?
* **Flexible Document Model:** Data is stored as BSON (Binary JSON) documents, which map naturally to objects in modern programming languages.
* **High Scalability:** Built-in sharding allows seamless horizontal scaling across distributed clusters.
* **High Availability:** Replica sets provide automatic failover and data redundancy.
* **Rich Query Language & Indexing:** Supports powerful querying, text search, geospatial queries, and secondary indexes for high performance.
* **Agility:** Schema changes can be made without downtime or complex database migrations, accelerating development cycles.

### 3. Explain the concept of collections in MongoDB.
* A collection is a grouping of MongoDB documents. It is the NoSQL equivalent of a relational database table.
* Collections exist within a database and do not enforce a rigid schema, meaning documents within the same collection can have varying fields and data types.

### 4. How does MongoDB ensure high availability using replication?
* MongoDB uses **Replica Sets**, which are groups of mongod processes that maintain the same data set.
* A replica set consists of one **Primary** node (which receives all write operations) and multiple **Secondary** nodes (which replicate the primary's oplog and apply the operations to their own datasets).
* If the primary node fails or becomes unreachable, the remaining secondary nodes automatically conduct an election to promote one of themselves to be the new primary, ensuring high availability with minimal downtime.

### 5. What are the main benefits of MongoDB Atlas?
* **Fully Managed:** Automated provisioning, patching, backups, and scaling without manual intervention.
* **Multi-Cloud Support:** Deploy clusters across AWS, Google Cloud, and Microsoft Azure.
* **Built-in Security:** Enterprise-grade security features including end-to-end encryption, VPC peering, IAM integration, and auditing.
* **Global Clusters:** Multi-region data distribution for low-latency access and compliance with data residency regulations.
* **Performance Advisor:** Automated index recommendations and performance monitoring tools.

### 6. What is the role of indexes in MongoDB, and how do they improve performance?
* Indexes support the efficient execution of queries in MongoDB. Without indexes, MongoDB must perform a *collection scan* (scanning every document in a collection) to find matching documents.
* Indexes store a small portion of the collection's data in an easy-to-traverse form (such as a B-tree), allowing MongoDB to locate and retrieve targeted documents quickly, dramatically reducing query execution time and CPU overhead.

### 7. Describe the stages of the MongoDB aggregation pipeline.
The aggregation pipeline processes documents through a sequence of stages to transform data into aggregated results. Common stages include:
* **`$match`**: Filters documents to pass only the documents that match specified conditions to the next stage.
* **`$group`**: Groups documents by a specified expression and applies accumulator expressions (like `$sum`, `$avg`).
* **`$project`**: Reshapes documents by adding, renaming, or removing fields.
* **`$sort`**: Sorts all input documents by specified sorting criteria.
* **`$limit` / `$skip`**: Restricts the number of documents passed or skips a given number of documents.
* **`$lookup`**: Performs a left outer join to another collection in the same database.

### 8. What is sharding in MongoDB? How does it differ from replication?
* **Sharding** is the process of distributing data across multiple machines (horizontal scaling) to handle growing data sizes and high throughput demands.
* **Difference from Replication:** Replication is about redundancy, fault tolerance, and high availability (copies of the same data across nodes), whereas sharding is about partitioning data across different servers so that each server holds a subset of the total data.

### 9. What is PyMongo, and why is it used?
* **PyMongo** is the official Python driver for MongoDB.
* It is used to connect Python applications to MongoDB databases, allowing developers to run queries, insert, update, delete documents, and manage database operations programmatically using Python code.

### 10. What are the ACID properties in the context of MongoDB transactions?
* **Atomicity:** All operations within a transaction succeed or all fail together.
* **Consistency:** Transactions bring the database from one valid state to another, maintaining schema validation rules.
* **Isolation:** Concurrent transactions do not interfere with each other; intermediate states are invisible to other operations.
* **Durability:** Once a transaction is committed, the changes are permanently saved and survive system crashes. MongoDB supports multi-document ACID transactions across replica sets and sharded clusters.

### 11. What is the purpose of MongoDB's `explain()` function?
* The `explain()` method provides information on the execution plan used by queries, updates, and deletes.
* It helps developers analyze query performance, check whether indexes are being utilized effectively, view the number of documents examined versus returned, and optimize slow-running queries.

### 12. How does MongoDB handle schema validation?
* MongoDB allows you to configure **JSON Schema validation** rules on collections using the `validator` option during collection creation or modification.
* When documents are inserted or updated, MongoDB checks them against the defined validation rules and rejects or warns if a document violates the required data types, required fields, or constraints.

### 13. What is the difference between a primary and a secondary node in a replica set?
* **Primary Node:** The only node in the replica set that accepts write operations and records all changes to its operations log (oplog).
* **Secondary Node:** Maintains a copy of the primary's data by replicating and applying the oplog. Secondaries can also handle read operations (if read preference allows) and participate in failover elections if the primary goes down.

### 14. What security mechanisms does MongoDB provide for data protection?
* **Authentication:** Role-Based Access Control (RBAC), LDAP, Kerberos, and x.509 certificates.
* **Authorization:** Granular privileges assigned to users and roles.
* **Encryption:** Encryption *at rest* (storage engine level) and encryption *in transit* using TLS/SSL.
* **Auditing:** Tracking security-relevant events and database activities.

### 15. Explain the concept of embedded documents and when they should be used.
* **Embedded documents** involve storing related data within a single document as nested sub-documents (denormalization).
* **When to use:** Use them for "has-one" or "has-many" relationships where data is frequently accessed together, and sub-documents do not grow indefinitely.

### 16. What is the purpose of MongoDB's `$lookup` stage in aggregation?
* The `$lookup` stage performs a left outer join from a collection in the same database to a collection in the same or different database, allowing you to combine related documents from two collections into a single result set.

### 17. What are some common use cases for MongoDB?
* Single View and Customer Data Platforms
* Content Management Systems (CMS) and E-commerce catalogs
* Real-time analytics and IoT data ingestion
* Mobile apps requiring flexible, rapidly evolving schemas
* Catalog and inventory management systems

### 18. What are the advantages of using MongoDB for horizontal scaling?
* **Sharding Architecture:** Automatically partitions data across multiple clusters without requiring application rewrites.
* **Elasticity:** Easily add or remove servers as load increases or decreases.
* **Cost Efficiency:** Scale out using commodity hardware rather than expensive vertical server upgrades.

### 19. How do MongoDB transactions differ from SQL transactions?
* While both support ACID guarantees, SQL transactions are standard across almost all relational database operations and tables. MongoDB transactions are designed primarily for complex multi-document updates across replica sets, while single-document operations are naturally atomic in MongoDB without needing explicit transactions.

### 20. What are the main differences between capped collections and regular collections?
* **Capped Collections:** Fixed-size circular collections that automatically overwrite their oldest documents when they reach their maximum size limit or document count limit. They preserve insertion order and provide ultra-fast reads/writes.
* **Regular Collections:** Dynamic size collections that grow as needed and require explicit deletion of old data.

### 21. What is the purpose of the `$match` stage in MongoDB's aggregation pipeline?
* The `$match` stage filters the document stream to pass only the documents that match specific query criteria to the subsequent pipeline stages, ideally placed at the beginning of the pipeline to reduce the dataset size early.

### 22. How can you secure access to a MongoDB database?
* Enable authentication (`auth = true` or via connection security settings).
* Create administrative users and enforce Role-Based Access Control (RBAC).
* Bind MongoDB to secure internal IP addresses and configure firewall rules / VPC security groups.
* Enable TLS/SSL encryption for network traffic.

### 23. What is MongoDB's WiredTiger storage engine, and why is it important?
* **WiredTiger** is the default storage engine in MongoDB.
* It is important because it provides advanced features like document-level concurrency control, checkpointing, memory-optimized cache management, and built-in block compression (snappy/zlib), which significantly improves performance and reduces storage space.

## Part 2: Practical Questions (Superstore Dataset)

### Environment Setup & MongoDB Connection

```python
# Install pymongo
!pip install pymongo pandas

import pandas as pd
import pymongo
from pymongo import MongoClient

# Connect to your MongoDB cluster (e.g. MongoDB Atlas or local mongod process)
# Replace connection string below with your Atlas URI if needed
client = MongoClient("mongodb://localhost:27017/")
db = client["superstore_db"]
orders_collection = db["Orders"]
print("Connected to MongoDB setup ready!")
```

### Q1. Write a Python script to load the Superstore dataset from a CSV file into MongoDB.

```python
# Load Superstore CSV into MongoDB
df = pd.read_csv('Superstore.csv', encoding='latin1')
data_records = df.to_dict(orient='records')
orders_collection.delete_many({})
orders_collection.insert_many(data_records)
print(f'Successfully loaded {orders_collection.count_documents({})} documents into MongoDB!')
```

### Q2. Retrieve and print all documents from the Orders collection.

```python
all_orders = list(orders_collection.find())
for doc in all_orders[:5]:  # Printing first 5 documents
    print(doc)
```

### Q3. Count and display the total number of documents in the Orders collection.

```python
total_docs = orders_collection.count_documents({})
print(f'Total number of documents in Orders collection: {total_docs}')
```

### Q4. Write a query to fetch all orders from the "West" region.

```python
west_orders = orders_collection.find({'Region': 'West'})
print(f"Total orders in West region: {orders_collection.count_documents({'Region': 'West'})}")
for order in west_orders.limit(3):
    print(order)
```

### Q5. Write a query to find orders where Sales is greater than 500.

```python
high_sales_orders = orders_collection.find({'Sales': {'$gt': 500}})
print(f"Total orders with Sales > 500: {orders_collection.count_documents({'Sales': {'$gt': 500}})}")
for order in high_sales_orders.limit(3):
    print(order)
```

### Q6. Fetch the top 3 orders with the highest Profit.

```python
top_3_profit = orders_collection.find().sort('Profit', pymongo.DESCENDING).limit(3)
print('Top 3 Orders by Profit:')
for order in top_3_profit:
    print(f"Order ID: {order.get('Order ID')} | Profit: ₹{order.get('Profit')} | Sales: ₹{order.get('Sales')}")
```

### Q7. Update all orders with Ship Mode as "First Class" to "Premium Class."

```python
update_result = orders_collection.update_many(
    {'Ship Mode': 'First Class'},
    {'$set': {'Ship Mode': 'Premium Class'}}
)
print(f'Matched documents: {update_result.matched_count}')
print(f'Modified documents: {update_result.modified_count}')
```

### Q8. Delete all orders where Sales is less than 50.

```python
delete_result = orders_collection.delete_many({'Sales': {'$lt': 50}})
print(f'Deleted documents where Sales < 50: {delete_result.deleted_count}')
```

### Q9. Use aggregation to group orders by Region and calculate total sales per region.

```python
pipeline = [
    {'$group': {'_id': '$Region', 'TotalSales': {'$sum': '$Sales'}}},
    {'$sort': {'TotalSales': -1}}
]
region_sales = orders_collection.aggregate(pipeline)
for result in region_sales:
    print(f"Region: {result['_id']} | Total Sales: ₹{result['TotalSales']:.2f}")
```

### Q10. Fetch all distinct values for Ship Mode from the collection.

```python
ship_modes = orders_collection.distinct('Ship Mode')
print('Distinct Ship Modes:', ship_modes)
```

### Q11. Count the number of orders for each category.

```python
category_pipeline = [
    {'$group': {'_id': '$Category', 'OrderCount': {'$sum': 1}}},
    {'$sort': {'OrderCount': -1}}
]
category_counts = orders_collection.aggregate(category_pipeline)
for cat in category_counts:
    print(f"Category: {cat['_id']} | Order Count: {cat['OrderCount']}")
```
